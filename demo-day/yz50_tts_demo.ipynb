# Sıfırdan Türkçe TTS (99M), benim sesimle

YZ50 Demo Day projesi. GPT-2 mimarisini sıfırdan yazdım, Türkçe konuşmada ön eğittim, sonra kendi ses kayıtlarımla ince ayar yaptım. Model metni Mimi codec token'larına çevirir, Mimi çözücü sesi üretir. Bu ses yapaydır.

**Çalıştırmak için:** üstten `Çalışma zamanı → Tümünü çalıştır` (Runtime → Run all). Aşağıda bir kutu açılır, cümleni yaz, **Konuştur**'a bas.
