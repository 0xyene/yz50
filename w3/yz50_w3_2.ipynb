```python
#Sayım tablosunu satır satır olasılıklara çevir ve modelden yeni isimler örnekle.
#Broadcasting'e dikkat: videodaki keepdim tuzağı, sessizce yanlış model üreten cinsten. 
import torch
words = open('names.txt', 'r').read().splitlines()
```

```python
N = torch.zeros(27,27, dtype=torch.int32)
```

```python
chars = sorted(list(set(''.join(words))))
stoi = {s:i+1 for i,s in enumerate(chars)}
stoi['.'] = 0
itos = {i:s for s,i in stoi.items()}
```

```python
for w in words:
    chs = ["."] + list(w) + ["."] 
    for ch1, ch2 in zip(chs,chs[1:]):
        ix1 = stoi[ch1]
        ix2 = stoi[ch2]
        N[ix1,ix2,] += 1 
```

```python
P = (N).float() #Could also add +1 to N -[N+1]- for giving a chance to that never occurred biagrams in data.
P = P/ P.sum(1, keepdims=True)
```

```python
P[0].sum(),P[1].sum() #We are cool. 
```

```python
#Due to torch version difference outputs are different from video although with same seed 
g = torch.Generator().manual_seed(2147483647)

for i in range(5):
  
  out = []
  ix = 0
  while True:
    p = P[ix]
    ix = torch.multinomial(p, num_samples=1, replacement=True, generator=g).item()
    out.append(itos[ix])
    if ix == 0:
      break
  print(''.join(out))
```

```python
P_wrong = (N).float() #Could also add +1 to N -[N+1]- for giving a chance to that never occurred biagrams in data.
P_wrong = P_wrong/ P_wrong.sum(1, keepdims=False)
```

```python
P_wrong[0].sum(),P_wrong[1].sum() #False normalizations' results. 
```
