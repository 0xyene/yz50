```python
#unicode code points for Turkish sentence
a = [ord(x) for x in "Çocuğun gözleri şüpheli görünüyordu"]
a
```

```python
b =list("Çocuğun gözleri şüpheli görünüyordu".encode('utf-8'))
b
```
