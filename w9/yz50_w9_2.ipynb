```python
with open("altayli_kose_yazilari.txt", "r", encoding = "utf-8") as f:
    text = f.read()

tokens = text.encode("utf-8")
tokens = list(map(int,tokens))
```

```python
#just like bigram week
def get_stats(ids):
    counts = {}
    for pair in zip(ids,ids[1:]):
        counts[pair] = counts.get(pair,0)+1
    return counts

stats = get_stats(tokens)
print(sorted(((v,k) for k,v in stats.items()), reverse=True))
```

```python
def merge (ids, pair, idx):
    newids = []
    i = 0
    while i < len(ids):
        if i < len(ids) - 1 and ids[i] == pair[0] and ids[i+1] == pair [1]:
            newids.append(idx)
            i += 2 
        else:
            newids.append(ids [i])
            i += 1
    return newids
```

```python
merges = {} 
ids = list(tokens) #copy of tokens
for i in range(50):
    stats = get_stats(ids)
    pair = max(stats, key=stats.get)
    idx = 256 + i
    print(f"merging {pair} into a new token {idx}")
    ids = merge(ids, pair, idx)
    merges[pair] = idx
```

```python
print(len(tokens)/len(ids)) #compression rate
```

```python
vocab = {idx: bytes([idx]) for idx in range(256)}
for (p0,p1), idx in merges.items():
    vocab[idx] = vocab[p0] + vocab[p1]

for idx in range(256, 306):
    print(idx, vocab[idx], vocab[idx].decode('utf-8', errors='replace'))
```
